```python
%sql

CREATE OR REPLACE TABLE silver.medicamento AS

SELECT
    id_medicamento,
    TRIM(nombre) AS nombre,
    TRIM(tipo) AS tipo,
    TRIM(fabricante) AS fabricante,
    CAST(stock AS INT) AS stock
FROM bronze.medicamento;
```

```python
%sql

SELECT *
FROM silver.medicamento
ORDER BY id_medicamento;
```

```python
%sql

SELECT *
FROM silver.medicamento
ORDER BY id_medicamento;
```

```python
%sql

SELECT COUNT(*) AS total
FROM silver.medicamento;
```

```python
%sql

CREATE OR REPLACE TABLE silver.openfda_medicamentos AS

SELECT
    id_medicamento,
    TRIM(nombre_medicamento) AS nombre_medicamento,
    TRIM(nombre_api) AS nombre_api,
    CAST(codigo_respuesta AS INT) AS codigo_respuesta,
    CAST(encontrado AS BOOLEAN) AS encontrado,
    raw_json
FROM bronze.openfda_medicamentos;
```

```python
%sql

SELECT *
FROM silver.openfda_medicamentos
ORDER BY id_medicamento;
```

```python
%sql

SELECT COUNT(*) AS total
FROM silver.openfda_medicamentos;
```
