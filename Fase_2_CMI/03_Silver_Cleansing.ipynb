```python
%sql

CREATE OR REPLACE TABLE silver.medicamento AS

SELECT
    id_medicamento,
    TRIM(nombre) AS nombre,
    TRIM(tipo) AS tipo,
    TRIM(fabricante) AS fabricante,
    CAST(stock AS INT) AS stock
FROM bronze.medicamento;
```

```python
%sql

SELECT *
FROM silver.medicamento
ORDER BY id_medicamento;
```

```python
%sql

CREATE OR REPLACE TABLE silver.openfda_medicamentos AS

SELECT
    id_medicamento,
    TRIM(nombre_medicamento) AS nombre_medicamento,
    TRIM(nombre_api) AS nombre_api,
    CAST(codigo_respuesta AS INT) AS codigo_respuesta,
    CAST(encontrado AS BOOLEAN) AS encontrado,
    raw_json
FROM bronze.openfda_medicamentos;
```

```python
%sql

SELECT *
FROM silver.openfda_medicamentos
ORDER BY id_medicamento;
```

```python
%sql

SELECT COUNT(*) AS total
FROM silver.openfda_medicamentos;
```

```python
%sql

CREATE OR REPLACE TABLE silver.rxnorm_medicamentos AS
SELECT
    TRIM(medicamento_cmi) AS medicamento_cmi,
    TRIM(rxcui) AS rxcui,
    TRIM(nombre_normalizado) AS nombre_normalizado
FROM bronze.rxnorm_medicamentos
WHERE medicamento_cmi IS NOT NULL
  AND TRIM(medicamento_cmi) <> '';

CREATE OR REPLACE TABLE silver.pubchem_medicamentos AS
SELECT
    TRIM(medicamento_cmi) AS medicamento_cmi,
    cid,
    TRIM(formula_molecular) AS formula_molecular,
    TRY_CAST(peso_molecular AS DOUBLE) AS peso_molecular,
    TRIM(inchikey) AS inchikey
FROM bronze.pubchem_medicamentos
WHERE medicamento_cmi IS NOT NULL
  AND TRIM(medicamento_cmi) <> '';

CREATE OR REPLACE TABLE silver.clinicaltrials_medicamentos AS
SELECT DISTINCT
    TRIM(medicamento_cmi) AS medicamento_cmi,
    TRIM(nct_id) AS nct_id,
    TRIM(titulo) AS titulo,
    TRIM(estado) AS estado,
    TRIM(tipo_estudio) AS tipo_estudio,
    TRIM(fase) AS fase,
    TRIM(condiciones) AS condiciones,
    TRIM(intervenciones) AS intervenciones
FROM bronze.clinicaltrials_medicamentos
WHERE medicamento_cmi IS NOT NULL
  AND TRIM(medicamento_cmi) <> ''
  AND nct_id IS NOT NULL;

CREATE OR REPLACE TABLE silver.who_icd11_diagnosticos AS
SELECT
    id_diagnostico,
    TRIM(diagnostico_cmi) AS diagnostico_cmi,
    TRIM(descripcion_cmi) AS descripcion_cmi,
    TRIM(nombre_who) AS nombre_who,
    TRIM(codigo_icd11) AS codigo_icd11,
    TRIM(id_who) AS id_who,
    score
FROM bronze.who_icd11_diagnosticos
WHERE diagnostico_cmi IS NOT NULL
  AND TRIM(diagnostico_cmi) <> '';
```

```python
%sql

SELECT 'RxNorm' AS fuente, COUNT(*) AS registros
FROM silver.rxnorm_medicamentos

UNION ALL

SELECT 'PubChem', COUNT(*)
FROM silver.pubchem_medicamentos

UNION ALL

SELECT 'ClinicalTrials', COUNT(*)
FROM silver.clinicaltrials_medicamentos

UNION ALL

SELECT 'OMS ICD-11', COUNT(*)
FROM silver.who_icd11_diagnosticos;
```
