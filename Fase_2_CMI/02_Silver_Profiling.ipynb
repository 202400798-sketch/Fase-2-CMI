```python
%sql
SELECT
    encontrado,
    COUNT(*) AS cantidad
FROM bronze.openfda_medicamentos
GROUP BY encontrado;
```

```python
%sql
SELECT
    COUNT(*) AS total,
    COUNT(id_medicamento) AS id_medicamento_no_nulos,
    COUNT(nombre_medicamento) AS nombre_no_nulo,
    COUNT(nombre_api) AS nombre_api_no_nulo,
    COUNT(codigo_respuesta) AS codigo_no_nulo,
    COUNT(encontrado) AS encontrado_no_nulo,
    COUNT(raw_json) AS raw_json_no_nulo
FROM bronze.openfda_medicamentos;
```

```python
%sql
SELECT
    id_medicamento,
    COUNT(*) AS cantidad
FROM bronze.openfda_medicamentos
GROUP BY id_medicamento
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT
    codigo_respuesta,
    COUNT(*) AS cantidad
FROM bronze.openfda_medicamentos
GROUP BY codigo_respuesta
ORDER BY codigo_respuesta;
```

```python
%sql
SELECT
    COUNT(*) AS total,
    COUNT(id_medicamento) AS id_medicamento_no_nulos,
    COUNT(nombre_medicamento) AS nombre_no_nulo,
    COUNT(nombre_api) AS nombre_api_no_nulo,
    COUNT(codigo_respuesta) AS codigo_no_nulo,
    COUNT(encontrado) AS encontrado_no_nulo,
    COUNT(raw_json) AS raw_json_no_nulo
FROM bronze.openfda_medicamentos;
```

```python
%sql
SELECT
    encontrado,
    COUNT(*) AS cantidad
FROM bronze.openfda_medicamentos
GROUP BY encontrado;
```

```python
%sql
SELECT
    COUNT(*) - COUNT(id_medicamento) AS nulos_id_medicamento,
    COUNT(*) - COUNT(nombre_medicamento) AS nulos_nombre_medicamento,
    COUNT(*) - COUNT(nombre_api) AS nulos_nombre_api,
    COUNT(*) - COUNT(codigo_respuesta) AS nulos_codigo_respuesta,
    COUNT(*) - COUNT(encontrado) AS nulos_encontrado,
    COUNT(*) - COUNT(raw_json) AS nulos_raw_json
FROM bronze.openfda_medicamentos;
```

```python
%sql
SELECT
    'RxNorm' AS fuente,
    COUNT(*) AS registros,
    COUNT(DISTINCT medicamento_cmi) AS medicamentos,
    SUM(CASE WHEN medicamento_cmi IS NULL THEN 1 ELSE 0 END) AS nombres_nulos
FROM bronze.rxnorm_medicamentos

UNION ALL

SELECT
    'PubChem',
    COUNT(*),
    COUNT(DISTINCT medicamento_cmi),
    SUM(CASE WHEN medicamento_cmi IS NULL THEN 1 ELSE 0 END)
FROM bronze.pubchem_medicamentos

UNION ALL

SELECT
    'ClinicalTrials',
    COUNT(*),
    COUNT(DISTINCT medicamento_cmi),
    SUM(CASE WHEN medicamento_cmi IS NULL THEN 1 ELSE 0 END)
FROM bronze.clinicaltrials_medicamentos

UNION ALL

SELECT
    'OMS ICD-11',
    COUNT(*),
    COUNT(DISTINCT diagnostico_cmi),
    SUM(CASE WHEN diagnostico_cmi IS NULL THEN 1 ELSE 0 END)
FROM bronze.who_icd11_diagnosticos;
```
