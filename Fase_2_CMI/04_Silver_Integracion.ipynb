```python
%sql

CREATE OR REPLACE TABLE silver.medicamentos_multifuente AS
SELECT
    m.id_medicamento,
    m.nombre,
    m.tipo,
    m.fabricante,
    m.stock,

    f.nombre_generico,
    f.nombre_comercial,
    f.ingrediente_activo,
    f.proposito,
    f.indicaciones_uso,
    f.advertencias,
    f.no_usar,
    f.consultar_medico,
    f.dosificacion,
    f.almacenamiento,
    f.fabricante_fda,
    f.tipo_producto,
    f.via_administracion,
    f.codigo_ndc,
    f.fecha_vigencia,

    r.rxcui,
    r.nombre_normalizado,

    p.cid,
    p.formula_molecular,
    p.peso_molecular,
    p.inchikey,

    CASE
        WHEN r.rxcui IS NOT NULL THEN 'Encontrado'
        ELSE 'No encontrado'
    END AS estado_rxnorm,

    CASE
        WHEN p.cid IS NOT NULL THEN 'Encontrado'
        ELSE 'No encontrado'
    END AS estado_pubchem

FROM silver.medicamento m

LEFT JOIN silver.medicamento_enriquecido_final f
    ON m.id_medicamento = f.id_medicamento

LEFT JOIN silver.rxnorm_medicamentos r
    ON LOWER(TRIM(m.nombre)) = LOWER(TRIM(r.medicamento_cmi))

LEFT JOIN silver.pubchem_medicamentos p
    ON LOWER(TRIM(m.nombre)) = LOWER(TRIM(p.medicamento_cmi));
```

```python
%sql

SELECT *
FROM silver.medicamentos_multifuente
ORDER BY id_medicamento;
```

```python
%sql

SELECT
    COUNT(*) AS total_registros,
    COUNT(DISTINCT id_medicamento) AS medicamentos_distintos,
    SUM(CASE WHEN rxcui IS NOT NULL THEN 1 ELSE 0 END)
        AS coincidencias_rxnorm,
    SUM(CASE WHEN cid IS NOT NULL THEN 1 ELSE 0 END)
        AS coincidencias_pubchem
FROM silver.medicamentos_multifuente;
```

```python
%sql
CREATE OR REPLACE TABLE silver.resumen_ensayos_medicamentos AS
SELECT
    TRIM(medicamento_cmi) AS medicamento_cmi,
    COUNT(DISTINCT nct_id) AS total_ensayos,
    SUM(CASE WHEN UPPER(estado) = 'RECRUITING'
        THEN 1 ELSE 0 END) AS ensayos_reclutando,
    SUM(CASE WHEN UPPER(estado) = 'COMPLETED'
        THEN 1 ELSE 0 END) AS ensayos_completados,
    COUNT(DISTINCT estado) AS estados_distintos
FROM silver.clinicaltrials_medicamentos
WHERE medicamento_cmi IS NOT NULL
GROUP BY TRIM(medicamento_cmi);
```

```python
%sql
CREATE OR REPLACE TABLE silver.diagnosticos_icd11 AS
SELECT
    id_diagnostico,
    TRIM(diagnostico_cmi) AS diagnostico_cmi,
    TRIM(descripcion_cmi) AS descripcion_cmi,
    TRIM(nombre_who) AS nombre_who,
    TRIM(codigo_icd11) AS codigo_icd11,
    TRIM(id_who) AS id_who,
    score,
    CASE
        WHEN codigo_icd11 IS NOT NULL
             AND TRIM(codigo_icd11) <> ''
        THEN 'Con código ICD-11'
        ELSE 'Sin código ICD-11'
    END AS estado_mapeo
FROM silver.who_icd11_diagnosticos
WHERE diagnostico_cmi IS NOT NULL
  AND TRIM(diagnostico_cmi) <> '';
```
