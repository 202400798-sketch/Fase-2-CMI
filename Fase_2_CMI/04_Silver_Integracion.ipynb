```python
%sql

CREATE OR REPLACE TABLE silver.medicamento_enriquecido AS

SELECT
    m.id_medicamento,
    m.nombre,
    m.tipo,
    m.fabricante,
    m.stock,
    f.nombre_api,
    f.codigo_respuesta,
    f.encontrado,
    f.raw_json
FROM silver.medicamento m
LEFT JOIN silver.openfda_medicamentos f
    ON m.id_medicamento = f.id_medicamento;
```

```python
%sql

CREATE OR REPLACE TABLE silver.medicamento_detalle AS

SELECT
    id_medicamento,
    nombre_medicamento,
    nombre_api,

    get_json_object(raw_json, '$.active_ingredient[0]') AS ingrediente_activo,

    get_json_object(raw_json, '$.purpose[0]') AS proposito,

    get_json_object(raw_json, '$.indications_and_usage[0]') AS indicaciones_uso,

    get_json_object(raw_json, '$.warnings[0]') AS advertencias,

    get_json_object(raw_json, '$.do_not_use[0]') AS no_usar,

    get_json_object(raw_json, '$.ask_doctor[0]') AS consultar_medico,

    get_json_object(raw_json, '$.dosage_and_administration[0]') AS dosificacion,

    get_json_object(raw_json, '$.storage_and_handling[0]') AS almacenamiento,

    get_json_object(raw_json, '$.inactive_ingredient[0]') AS ingredientes_inactivos,

    get_json_object(raw_json, '$.effective_time') AS fecha_vigencia,

    get_json_object(raw_json, '$.openfda.generic_name[0]') AS nombre_generico,

    get_json_object(raw_json, '$.openfda.brand_name[0]') AS nombre_comercial,

    get_json_object(raw_json, '$.openfda.manufacturer_name[0]') AS fabricante_fda,

    get_json_object(raw_json, '$.openfda.product_type[0]') AS tipo_producto,

    get_json_object(raw_json, '$.openfda.route[0]') AS via_administracion,

    get_json_object(raw_json, '$.openfda.product_ndc[0]') AS codigo_ndc

FROM silver.openfda_medicamentos;
```

```python
%sql
CREATE OR REPLACE TABLE silver.medicamento_detalle AS
SELECT
    id_medicamento,
    nombre_medicamento,
    nombre_api,

    get_json_object(raw_json, '$.active_ingredient[0]') AS ingrediente_activo,
    get_json_object(raw_json, '$.purpose[0]') AS proposito,
    get_json_object(raw_json, '$.indications_and_usage[0]') AS indicaciones_uso,
    get_json_object(raw_json, '$.warnings[0]') AS advertencias,
    get_json_object(raw_json, '$.do_not_use[0]') AS no_usar,
    get_json_object(raw_json, '$.ask_doctor[0]') AS consultar_medico,
    get_json_object(raw_json, '$.dosage_and_administration[0]') AS dosificacion,
    get_json_object(raw_json, '$.storage_and_handling[0]') AS almacenamiento,
    get_json_object(raw_json, '$.inactive_ingredient[0]') AS ingredientes_inactivos,

    get_json_object(raw_json, '$.effective_time') AS fecha_vigencia,

    get_json_object(raw_json, '$.openfda.generic_name[0]') AS nombre_generico,
    get_json_object(raw_json, '$.openfda.brand_name[0]') AS nombre_comercial,
    get_json_object(raw_json, '$.openfda.manufacturer_name[0]') AS fabricante_fda,
    get_json_object(raw_json, '$.openfda.product_type[0]') AS tipo_producto,
    get_json_object(raw_json, '$.openfda.route[0]') AS via_administracion,
    get_json_object(raw_json, '$.openfda.product_ndc[0]') AS codigo_ndc

FROM silver.openfda_medicamentos;
```

```python
%sql
SELECT
    COUNT(*) AS total_registros,

    COUNT(ingrediente_activo) AS ingrediente_activo_completos,
    COUNT(proposito) AS proposito_completos,
    COUNT(indicaciones_uso) AS indicaciones_completos,
    COUNT(advertencias) AS advertencias_completos,
    COUNT(no_usar) AS no_usar_completos,
    COUNT(dosificacion) AS dosificacion_completos,
    COUNT(almacenamiento) AS almacenamiento_completos,
    COUNT(nombre_generico) AS nombre_generico_completos,
    COUNT(fabricante_fda) AS fabricante_fda_completos,
    COUNT(via_administracion) AS via_administracion_completos,
    COUNT(codigo_ndc) AS codigo_ndc_completos

FROM silver.medicamento_detalle;
```
