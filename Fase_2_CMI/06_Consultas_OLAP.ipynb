```python
%sql
SELECT
    nombre AS medicamento,
    tipo,
    COUNT(DISTINCT id_receta) AS total_recetas,
    SUM(cantidad) AS unidades_prescritas,
    ROUND(AVG(cantidad), 2) AS promedio_unidades_por_receta
FROM gold.fact_prescripciones
GROUP BY
    nombre,
    tipo
ORDER BY
    unidades_prescritas DESC;
```

```python
%sql
SELECT
    nombre AS medicamento,
    tipo,
    MAX(stock) AS stock_actual,
    SUM(cantidad) AS unidades_prescritas,

    ROUND(
        SUM(cantidad) / NULLIF(MAX(stock), 0),
        2
    ) AS relacion_demanda_stock,

    CASE
        WHEN SUM(cantidad) / NULLIF(MAX(stock), 0) >= 2
            THEN 'Alta'
        WHEN SUM(cantidad) / NULLIF(MAX(stock), 0) >= 1
            THEN 'Media'
        ELSE 'Baja'
    END AS nivel_demanda_stock

FROM gold.fact_prescripciones

GROUP BY
    nombre,
    tipo

ORDER BY
    relacion_demanda_stock DESC;
```

```python
%sql
SELECT
    tipo,
    COUNT(DISTINCT id_receta) AS total_recetas,
    COUNT(DISTINCT id_medicamento) AS medicamentos_diferentes,
    SUM(cantidad) AS unidades_prescritas,
    ROUND(AVG(cantidad), 2) AS promedio_unidades_por_detalle
FROM gold.fact_prescripciones
GROUP BY
    tipo
ORDER BY
    unidades_prescritas DESC;
```

```python
%sql
SELECT
    YEAR(fecha_receta) AS anio,
    MONTH(fecha_receta) AS mes,
    COUNT(DISTINCT id_receta) AS total_recetas,
    COUNT(DISTINCT id_medicamento) AS medicamentos_diferentes,
    SUM(cantidad) AS unidades_prescritas,
    ROUND(AVG(cantidad), 2) AS promedio_unidades
FROM gold.fact_prescripciones
GROUP BY
    YEAR(fecha_receta),
    MONTH(fecha_receta)
ORDER BY
    anio,
    mes;
```

```python
%sql
SELECT
    nombre,
    tipo,
    stock,
    CASE
        WHEN stock < 400 THEN 'Stock bajo'
        WHEN stock < 600 THEN 'Stock medio'
        ELSE 'Stock suficiente'
    END AS nivel_stock
FROM gold.medicamentos_multifuente
ORDER BY stock ASC;
```

```python
%sql
SELECT
    estado_rxnorm,
    estado_pubchem,
    COUNT(*) AS cantidad_medicamentos
FROM gold.medicamentos_multifuente
GROUP BY estado_rxnorm, estado_pubchem
ORDER BY cantidad_medicamentos DESC;
```

```python
%sql
SELECT
    medicamento_cmi,
    total_ensayos,
    ensayos_reclutando,
    ensayos_completados
FROM gold.resumen_ensayos_medicamentos
ORDER BY total_ensayos DESC
LIMIT 10;
```

```python
%sql
SELECT
    estado_mapeo,
    COUNT(*) AS cantidad,
    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (),
        2
    ) AS porcentaje
FROM gold.resumen_diagnosticos_icd11
GROUP BY estado_mapeo;
```

```python
%sql
SELECT
    tipo,
    COUNT(*) AS medicamentos_distintos,
    SUM(stock) AS unidades_disponibles,
    ROUND(AVG(stock), 2) AS stock_promedio
FROM gold.medicamentos_multifuente
GROUP BY tipo
ORDER BY unidades_disponibles DESC;
```
